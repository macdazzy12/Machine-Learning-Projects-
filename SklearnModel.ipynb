```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
from sklearn.impute import SimpleImputer
from sklearn.inspection import permutation_importance
import statsmodels.api as sm
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv(r"C:\Users\HP\Downloads\1.01. Simple linear regression Sklearn.csv")
```

```python
data.head()
```

```python
y= data["GPA"]
x= data["SAT"]
```

```python
x.shape
```

```python
y.shape
```

```python
x_matrix = x.values.reshape(-1,1)
x_matrix.shape
```

```python
reg= LinearRegression()
```

```python
reg.fit(x_matrix,y)
```

```python
reg.score(x_matrix,y)
```

```python
reg.coef_
#the result will be an nd array containing all coeeficients in this case there is a single intercept
```

```python
reg.intercept_
#we will get a float instead of an array because this type of regression always has a single intercept
```

```python
reg.predict([[1740]])
```

```python
values = [[1740], [1800], [2000]]
reg.predict(values)
#here we are predicting for multiple point values 
```

```python
plt.scatter(x,y)
plt.xlabel("SAT", fontsize=20)
plt.ylabel("GPA", fontsize=20)
plt.show()
```

```python
y_pred = reg.predict(x)
plt.scatter(x, y, color='blue', label='Actual data')
plt.plot(x, y_pred, color='red', linewidth=2, label='Regression line')
plt.xlabel('SAT')
plt.ylabel('GPA')
plt.legend()
plt.show()
```
